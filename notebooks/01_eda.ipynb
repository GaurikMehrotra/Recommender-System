```python
import os
from pathlib import Path

root = Path.cwd()
if root.name == "notebooks":
    os.chdir(root.parent)
```

```python
from IPython.display import Image, Markdown, display

from src.data.preprocess import load_movielens
from src.report import eda

eda("data/raw/ml-100k", Path("reports/figures"))
ratings, items, features = load_movielens("data/raw/ml-100k")
```

```python
sparsity = 1 - len(ratings) / (ratings.user.nunique() * len(items))
display(
    Markdown(
        f"**Takeaway:** {sparsity:.2%} of user-catalog pairs are unobserved, so sparse storage is appropriate."
    )
)
```

```python
display(Image("reports/figures/popularity.png"))
counts = ratings.groupby("item").size().sort_values(ascending=False)
head = max(1, int(len(items) * 0.2))
share = counts.iloc[:head].sum() / counts.sum()
display(
    Markdown(
        f"**Takeaway:** the most-rated 20% of the catalog accounts for {share:.1%} of observed ratings, revealing exposure concentration."
    )
)
```

```python
display(Image("reports/figures/activity.png"))
activity = ratings.groupby("user").size()
display(
    Markdown(
        f"**Takeaway:** median activity is {activity.median():.0f} ratings, compared with a maximum of {activity.max()}, so macro user averaging limits dominance by active users."
    )
)
```

```python
display(Image("reports/figures/ratings.png"))
display(
    Markdown(
        f"**Takeaway:** {(ratings.rating >= 4).mean():.1%} of ratings meet the configured positive threshold; observed ratings are not unbiased implicit exposure."
    )
)
```
