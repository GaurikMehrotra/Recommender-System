# Bias and cold-start diagnostics
Cold-start labels are validation only; fixed reranking curves are descriptive test analyses.

```python
import os
from pathlib import Path

root = Path.cwd()
if root.name == "notebooks":
    os.chdir(root.parent)
```

```python
import pandas as pd
from IPython.display import Markdown, display

frame = pd.read_csv("runs/ml100k/results.csv")
columns = ["model", "coverage", "novelty_bits", "intra_list_diversity", "head_share", "NDCG@20"]
display(frame[columns].groupby("model").mean())
```

```python
import matplotlib.pyplot as plt

ax = (
    frame.groupby("model")[["head_share", "NDCG@20"]]
    .mean()
    .plot.scatter(x="head_share", y="NDCG@20")
)
plt.show()
display(
    Markdown(
        "**Takeaway:** interpret the observed concentration/accuracy tradeoff without selecting a best reranking strength on test."
    )
)
```

```python
cold = pd.read_csv("runs/ml100k_cold/cold_start.csv")
display(cold)
users = cold[cold["mode"] == "users"]
users.pivot(index="history_size", columns="model", values="NDCG@20").plot(marker="o")
plt.ylabel("Validation NDCG@20")
plt.show()
display(
    Markdown(
        "**Takeaway:** compare the measured 1-5-history curves under matched cohorts; synthetic cold starts are not proof of real new-user behavior."
    )
)
```
